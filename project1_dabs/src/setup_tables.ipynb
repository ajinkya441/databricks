```python
dbutils.widgets.text("catalog", "workspace")
dbutils.widgets.text("schema", "schema_dev")
dbutils.widgets.text("environment", "dev")

v_catalog = dbutils.widgets.get("catalog")
v_schema = dbutils.widgets.get("schema")
v_environment = dbutils.widgets.get("environment")
```

```python
sql_code = f"CREATE CATALOG IF NOT EXISTS {v_catalog}"
print(sql_code)
spark.sql(sql_code)
```

```python
# Then create tables
sql_code2 = f"CREATE SCHEMA IF NOT EXISTS {v_catalog}.{v_schema}"
print(sql_code2)
spark.sql(sql_code2)

sql_code3 = f"""
CREATE OR REPLACE TABLE {v_catalog}.{v_schema}.nyctaxi
DEEP CLONE samples.nyctaxi.trips
"""
print(sql_code3)
spark.sql(sql_code3)
```
